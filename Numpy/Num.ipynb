# Numpy Arrays Vs Python List
- Numpy Arrays are faster (Optimized in C)
- Numpy Stores Data in Continous Memory Block
- Uses Less Memory (optimized Storage)
- Built-in Mathematical Operations
- Supports Direct Vectorized Operations 

```python
print("At 1st going to do square of both python list and numpy array, and finding execution time for both")
size = 1_000_000
#python list
py_list = list(range(size))
start = time.time()
sq_list = [x ** 2 for x in py_list]
end = time.time()
print (f"Python List time= {end - start} seconds")

#numpy array
np_arr = np.array(py_list)
start = time.time()
sq_array = np_arr ** 2
end = time.time()
print(f"Numpy array time {end - start} seconds")
```

```python
#memory
import sys 
print(f"Python List size = {sys.getsizeof(py_list) * len(py_list)} bytes")
print(f"Numpy Array size =  {np_arr.nbytes} bytes")
```

# Operations on Numpy Array

```python
# Reshaping (2,3) -> (3,2)
arr = np.array([[1,2,3],[4,5,6]])
print(arr, arr.shape)

reshaped = arr.reshape((3,2))
print(reshaped, reshaped.shape)

# Flatterned
flat = arr.flatten()
print(flat, flat.shape)
```

```python
# Indexing in 1D & 2D
# 1D
arr = np.array([1,2,3,4,5,6])
print(arr[0])
print(arr[1])
```

```python
# 2D
arr = np.array([[1,2,3],[4,5,6]])
print(arr[0][1])
```

```python
# Fancy Indexing
arr = np.array([1,2,3,4,5,6])
idx = [0,1,3]
print(arr[idx])
```

```python
# Boolean Indexing
arr = np.array([1,2,3,4,5,6])
print(arr[arr>2])
print(arr[arr%2 == 0])
print(arr[arr% 2!= 0 ])
```

```python
# Slicing (start, end, step)
arr = np.array([1,2,3,4,5])
print(arr[1:4])
print(arr[1:])
print(arr[:4])
print(arr[::2])
```

## Vector Normalization
-Normalizing an array means transdorming its values so they fit into a specific scale or range

```python
arr = np.array([[1, 2], [3, 4]])

mean = np.mean(arr)
std_dev = np.std(arr)

print("Mean:", mean)
print("Standard Deviation:", std_dev)

normalized_arr = (arr - mean) / std_dev

print("Normalized Array:")
print(normalized_arr)
```

# Mathematical Functions
## Aggregation
- Sum() - prod() - min() -max() -mean() -std() -var() - median() -argmin() - argmax() 
## Power
- square() - sqrt() - pow(a,b) 
## Log & Exponential
- log() -log10() - log2() - exp()
## Rounding
- round() - floor() - cell() - trunc()
## Extras
- unique() - sort() - abs()

```python
# Aggregation Functions
arr = np.array([1,2,3,4,5,6])
print(np.sum(arr))   # Sum of all Elements
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))  # Average
print(np.std(arr))
print(np.var(arr))
print(np.median(arr))  # Middle value
print(np.argmin(arr))  # Index of minimum value
print(np.argmax(arr))  # Index of maximum value
```

```python
# Power
arr = np.array([1,2,3,4,5,6])
print(np.square(arr))
print(np.sqrt(arr))   # Find Square Root for each element

a = np.array([2, 3, 4])
b = 2

print(np.pow(a, b))
```

```python
# Log & Exponential
arr = np.array([1, np.e, np.e**2])
print(np.log(arr))

arr2 = np.array([1, 10, 100, 1000])
print(np.log10(arr2))

arr3 = np.array([1, 2, 4, 8, 16])
print(np.log2(arr3))

arr4 = np.array([0, 1, 2, 3])
print(np.exp(arr4))
```

```python
# Rounding Functions
arr = np.array([1.2,2.6,2.4,8.2,8.9])
print(np.round(arr))

# Can Specify Decimal Spaces
arr2 = np.array([1.2345, 2.6789])
print(np.round(arr2, 2))

# np.floor round down, Floor means the greatest integer ≤ the number.
arr3 = np.array([1.2, 2.9, 3.1, -2.3])
print(np.floor(arr3))

# np.ceil() — Round up 
arr4 = np.array([1.2, 2.1, 3.9, -2.3])
print(np.ceil(arr4))

# np.trunc() — Remove decimal part
arr5 = np.array([1.9, 2.7, -3.8, -4.2])
print(np.trunc(arr5))
```

```python
# Extra Functions
# np.unique() — Find unique values , Duplicate Values are removed
arr = np.array([10, 20, 10, 30, 20, 40, 30])
print(np.unique(arr))

# np.sort() — Sort elements
arr2 = np.array([50, 10, 40, 20, 30])
print(np.sort(arr2))

# for Decending 
arr3 = np.array([50, 10, 40, 20, 30])
print(np.sort(arr3)[::-1])

# np.abs() — Absolute value, Removes Negative Signs
arr4 = np.array([-10, -5, 0, 5, 10])
print(np.abs(arr4))
```
