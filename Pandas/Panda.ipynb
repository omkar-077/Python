### Data Frame Methods 
- df.head()  # prints data of 1st 5 rows
- df.tail()  # last 5
- df.sample() # random row
- df.info()
- df.shape()
- df.describe()
- df.coloums()
- df.unique()

```python
df = pd.read_csv("Toyota.csv")

print(df.head())
print(df.tail())
print(df.sample())
print(df.info())
print(df.shape)
print(df.describe())
print(df.columns)
#print(df.unique)
```
