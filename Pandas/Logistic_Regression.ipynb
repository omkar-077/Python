```python
 
# -*- coding: utf-8 -*-
"""
Personal Income Classification using Logistic Regression
"""

# =============================================================================
# IMPORT REQUIRED LIBRARIES
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix


# =============================================================================
# SET VISUALIZATION STYLE
# =============================================================================

sns.set_theme(style="whitegrid")


# =============================================================================
# IMPORT DATA
# =============================================================================

# Read original dataset
data_income = pd.read_csv("income.csv")

# Create a copy of the original dataset
data = data_income.copy()


# =============================================================================
# EXPLORATORY DATA ANALYSIS
# =============================================================================

# -----------------------------------------------------------------------------
# 1. Dataset information
# -----------------------------------------------------------------------------

print("\n========== DATA INFORMATION ==========\n")
data.info()


# -----------------------------------------------------------------------------
# 2. Missing values
# -----------------------------------------------------------------------------

print("\n========== MISSING VALUES ==========\n")
print(data.isnull().sum())


# -----------------------------------------------------------------------------
# 3. Numerical variables summary
# -----------------------------------------------------------------------------

print("\n========== NUMERICAL SUMMARY ==========\n")
print(data.describe())


# -----------------------------------------------------------------------------
# 4. Categorical variables summary
# -----------------------------------------------------------------------------

print("\n========== CATEGORICAL SUMMARY ==========\n")
print(data.describe(include="object"))


# -----------------------------------------------------------------------------
# 5. Frequency of categories
# -----------------------------------------------------------------------------

print("\n========== JOB TYPE ==========\n")
print(data["JobType"].value_counts(dropna=False))

print("\n========== OCCUPATION ==========\n")
print(data["occupation"].value_counts(dropna=False))


# -----------------------------------------------------------------------------
# 6. Unique categories
# -----------------------------------------------------------------------------

print("\n========== UNIQUE JOB TYPES ==========\n")
print(np.unique(data["JobType"].dropna()))

print("\n========== UNIQUE OCCUPATIONS ==========\n")
print(np.unique(data["occupation"].dropna()))


# =============================================================================
# HANDLE " ?" AS MISSING VALUE
# =============================================================================

# Reload data and treat " ?" as NaN
data = pd.read_csv(
    "income.csv",
    na_values=[" ?"]
)

print("\n========== MISSING VALUES AFTER CLEANING ==========\n")
print(data.isnull().sum())


# =============================================================================
# DATA PRE-PROCESSING
# =============================================================================

# Rows containing at least one missing value
missing = data[data.isnull().any(axis=1)]

print("\nNumber of rows containing missing values:", len(missing))


# Remove rows containing missing values
data2 = data.dropna(axis=0).copy()

# Additional copies for separate models
data3 = data2.copy()
data4 = data2.copy()

print("\nOriginal dataset shape:", data.shape)
print("Cleaned dataset shape:", data2.shape)


# =============================================================================
# CROSS TABLES AND DATA VISUALIZATION
# =============================================================================


# -----------------------------------------------------------------------------
# 1. Gender proportion
# -----------------------------------------------------------------------------

gender = pd.crosstab(
    index=data2["gender"],
    columns="count",
    normalize=True
)

print("\n========== GENDER PROPORTION ==========\n")
print(gender)


# -----------------------------------------------------------------------------
# 2. Gender vs Salary Status
# -----------------------------------------------------------------------------

gender_salstat = pd.crosstab(
    index=data2["gender"],
    columns=data2["SalStat"],
    margins=True,
    normalize="index"
)

print("\n========== GENDER VS SALARY STATUS ==========\n")
print(gender_salstat)


# =============================================================================
# VISUALIZATIONS
# =============================================================================


# -----------------------------------------------------------------------------
# 3. Salary Status Distribution
# -----------------------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=data2,
    x="SalStat"
)

plt.title("Salary Status Distribution")
plt.xlabel("Salary Status")
plt.ylabel("Number of People")

plt.xticks(
    [0, 1],
    ["<= 50,000", "> 50,000"]
)

plt.tight_layout()
plt.show()


# -----------------------------------------------------------------------------
# 4. Age Distribution
# -----------------------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=data2,
    x="age",
    bins=10,
    kde=False
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of People")

plt.tight_layout()
plt.show()


# -----------------------------------------------------------------------------
# 5. Age vs Salary Status
# -----------------------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data2,
    x="SalStat",
    y="age"
)

plt.title("Age vs Salary Status")
plt.xlabel("Salary Status")
plt.ylabel("Age")

plt.xticks(
    [0, 1],
    ["<= 50,000", "> 50,000"]
)

plt.tight_layout()
plt.show()


# Median age by salary status
print("\n========== MEDIAN AGE BY SALARY STATUS ==========\n")
print(data2.groupby("SalStat")["age"].median())


# -----------------------------------------------------------------------------
# 6. Job Type vs Salary Status
# -----------------------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.countplot(
    data=data2,
    y="JobType",
    hue="SalStat"
)

plt.title("Job Type vs Salary Status")
plt.xlabel("Number of People")
plt.ylabel("Job Type")

plt.legend(
    title="Salary Status",
    labels=["<= 50,000", "> 50,000"]
)

plt.tight_layout()
plt.show()


# Job type cross table
job_salstat = pd.crosstab(
    index=data2["JobType"],
    columns=data2["SalStat"],
    margins=True,
    normalize="index"
)

print("\n========== JOB TYPE VS SALARY (%) ==========\n")
print(round(job_salstat * 100, 1))


# -----------------------------------------------------------------------------
# 7. Education Type vs Salary Status
# -----------------------------------------------------------------------------

plt.figure(figsize=(11, 6))

sns.countplot(
    data=data2,
    y="EdType",
    hue="SalStat"
)

plt.title("Education Type vs Salary Status")
plt.xlabel("Number of People")
plt.ylabel("Education Type")

plt.legend(
    title="Salary Status",
    labels=["<= 50,000", "> 50,000"]
)

plt.tight_layout()
plt.show()


# Education cross table
EdType_salstat = pd.crosstab(
    index=data2["EdType"],
    columns=data2["SalStat"],
    margins=True,
    normalize="index"
)

print("\n========== EDUCATION VS SALARY (%) ==========\n")
print(round(EdType_salstat * 100, 1))


# -----------------------------------------------------------------------------
# 8. Occupation vs Salary Status
# -----------------------------------------------------------------------------

plt.figure(figsize=(11, 7))

sns.countplot(
    data=data2,
    y="occupation",
    hue="SalStat"
)

plt.title("Occupation vs Salary Status")
plt.xlabel("Number of People")
plt.ylabel("Occupation")

plt.legend(
    title="Salary Status",
    labels=["<= 50,000", "> 50,000"]
)

plt.tight_layout()
plt.show()


# Occupation cross table
occ_salstat = pd.crosstab(
    index=data2["occupation"],
    columns=data2["SalStat"],
    margins=True,
    normalize="index"
)

print("\n========== OCCUPATION VS SALARY (%) ==========\n")
print(round(occ_salstat * 100, 1))


# -----------------------------------------------------------------------------
# 9. Capital Gain Distribution
# -----------------------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=data2,
    x="capitalgain",
    bins=10,
    kde=False
)

plt.title("Capital Gain Distribution")
plt.xlabel("Capital Gain")
plt.ylabel("Number of People")

plt.tight_layout()
plt.show()


# -----------------------------------------------------------------------------
# 10. Capital Loss Distribution
# -----------------------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=data2,
    x="capitalloss",
    bins=10,
    kde=False
)

plt.title("Capital Loss Distribution")
plt.xlabel("Capital Loss")
plt.ylabel("Number of People")

plt.tight_layout()
plt.show()


# =============================================================================
# LOGISTIC REGRESSION
# =============================================================================


# -----------------------------------------------------------------------------
# Convert Salary Status into 0 and 1
# -----------------------------------------------------------------------------

data2["SalStat"] = data2["SalStat"].map({
    " less than or equal to 50,000": 0,
    " greater than 50,000": 1
})

print("\n========== SALARY STATUS ==========\n")
print(data2["SalStat"].value_counts())


# -----------------------------------------------------------------------------
# One-hot encoding
# -----------------------------------------------------------------------------

new_data = pd.get_dummies(
    data2,
    drop_first=True
)


# -----------------------------------------------------------------------------
# Separate features and target
# -----------------------------------------------------------------------------

features = new_data.columns.tolist()
features.remove("SalStat")

y = new_data["SalStat"].values
x = new_data[features].values


print("\nNumber of features:", len(features))


# =============================================================================
# TRAIN-TEST SPLIT
# =============================================================================

train_x, test_x, train_y, test_y = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=0
)


print("\nTraining data:", train_x.shape)
print("Testing data:", test_x.shape)


# =============================================================================
# LOGISTIC REGRESSION MODEL
# =============================================================================

logistic = LogisticRegression(
    max_iter=1000
)


# Train model
logistic.fit(
    train_x,
    train_y
)


# Model coefficients
print("\n========== LOGISTIC COEFFICIENTS ==========\n")
print(logistic.coef_)

print("\n========== LOGISTIC INTERCEPT ==========\n")
print(logistic.intercept_)


# =============================================================================
# PREDICTION
# =============================================================================

prediction = logistic.predict(test_x)


# =============================================================================
# MODEL EVALUATION
# =============================================================================

# Confusion matrix
confusion_matrix_result = confusion_matrix(
    test_y,
    prediction
)

print("\n========== CONFUSION MATRIX ==========\n")
print(confusion_matrix_result)


# Accuracy
accuracy = accuracy_score(
    test_y,
    prediction
)

print("\nAccuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")


# Misclassified samples
misclassified = (test_y != prediction).sum()

print("\nMisclassified samples:", misclassified)


# =============================================================================
# CONFUSION MATRIX VISUALIZATION
# =============================================================================

plt.figure(figsize=(6, 5))

sns.heatmap(
    confusion_matrix_result,
    annot=True,
    fmt="d",
    xticklabels=["<= 50,000", "> 50,000"],
    yticklabels=["<= 50,000", "> 50,000"]
)

plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted Salary")
plt.ylabel("Actual Salary")

plt.tight_layout()
plt.show()


# =============================================================================
# LOGISTIC REGRESSION - REDUCED VARIABLES
# =============================================================================


# Create a fresh copy
data3 = data2.copy()


# -----------------------------------------------------------------------------
# Remove selected variables
# -----------------------------------------------------------------------------

cols_to_remove = [
    "gender",
    "nativecountry",
    "race",
    "JobType"
]

data3 = data3.drop(
    columns=cols_to_remove
)


# -----------------------------------------------------------------------------
# One-hot encoding
# -----------------------------------------------------------------------------

new_data2 = pd.get_dummies(
    data3,
    drop_first=True
)


# -----------------------------------------------------------------------------
# Separate features and target
# -----------------------------------------------------------------------------

features2 = new_data2.columns.tolist()
features2.remove("SalStat")

y2 = new_data2["SalStat"].values
x2 = new_data2[features2].values


print("\n========== REDUCED MODEL ==========\n")
print("Number of features:", len(features2))


# =============================================================================
# TRAIN-TEST SPLIT
# =============================================================================

train_x2, test_x2, train_y2, test_y2 = train_test_split(
    x2,
    y2,
    test_size=0.3,
    random_state=0
)


# =============================================================================
# SECOND LOGISTIC REGRESSION MODEL
# =============================================================================

logistic2 = LogisticRegression(
    max_iter=1000
)


# Train model
logistic2.fit(
    train_x2,
    train_y2
)


# Prediction
prediction2 = logistic2.predict(
    test_x2
)


# =============================================================================
# SECOND MODEL EVALUATION
# =============================================================================

confusion_matrix_result2 = confusion_matrix(
    test_y2,
    prediction2
)

accuracy2 = accuracy_score(
    test_y2,
    prediction2
)

misclassified2 = (test_y2 != prediction2).sum()


print("\n========== REDUCED MODEL RESULTS ==========\n")

print("Confusion Matrix:")
print(confusion_matrix_result2)

print("\nAccuracy:", accuracy2)
print("Accuracy percentage:", accuracy2 * 100, "%")

print("\nMisclassified samples:", misclassified2)


# =============================================================================
# SECOND CONFUSION MATRIX
# =============================================================================

plt.figure(figsize=(6, 5))

sns.heatmap(
    confusion_matrix_result2,
    annot=True,
    fmt="d",
    xticklabels=["<= 50,000", "> 50,000"],
    yticklabels=["<= 50,000", "> 50,000"]
)

plt.title("Reduced Logistic Regression Confusion Matrix")
plt.xlabel("Predicted Salary")
plt.ylabel("Actual Salary")

plt.tight_layout()
plt.show()


# =============================================================================
# MODEL COMPARISON
# =============================================================================

print("\n========== MODEL COMPARISON ==========\n")

print("Original Logistic Regression Accuracy:",
      round(accuracy * 100, 2), "%")

print("Reduced Logistic Regression Accuracy:",
      round(accuracy2 * 100, 2), "%")

print("\nOriginal Model Misclassified:",
      misclassified)

print("Reduced Model Misclassified:",
      misclassified2)


# =============================================================================
# END OF SCRIPT
# =============================================================================
 
```
